```python
xdim=np.shape(cube)[1]
ydim=np.shape(cube)[2]
nimg=np.shape(cube)[0]
npix=xdim*ydim
moy=[]
sig=[]
for i in range(xdim):
    for j in range(ydim):
        moy.append(np.mean(cube[:,i,j]))
        sig.append(np.std(cube[:,i,j]))
plt.plot(range(npix),sig,'.')
plt.yscale('log')
```

```python
x=range(npix)
for i in range(nimg):
    plt.plot(x,cube[i,:,:].flatten(),'.',ms=0.75)
#plt.yscale('log')
```

```python
plt.errorbar(x,np.abs(moy),yerr=sig,color="black",fmt='.',ms=1.25,linestyle='',elinewidth=0.75)
plt.yscale('log')
plt.ylim(0.1,200)
```

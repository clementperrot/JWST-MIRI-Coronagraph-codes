```python
import glob, os
from astropy.io import fits
import shutil
```

```python
def move_file_JWST(path,path_i):

    files = glob.glob(path+path_i+'*.fits',recursive=True)
    files.sort()
    print(str(len(files))+" files to move.")

    for fi in files:
        img = fits.open(fi)
        obs_nb = img[0].header['OBSERVTN']
        exp_ty = img[0].header['EXP_TYPE']
        path_dest = path+"obs"+str(obs_nb)+"/"
        if not os.path.exists(path_dest):
            os.makedirs(path_dest)
        if exp_ty == "MIR_TACQ":
            path_dest = path_dest+"TA/"
            if not os.path.exists(path_dest):
                os.makedirs(path_dest)
        if exp_ty == "NRC_TACQ":
            path_dest = path_dest+"TA/"
            if not os.path.exists(path_dest):
                os.makedirs(path_dest)
        shutil.move(fi, path_dest)
        

def version_fits_JWST(path):
    files = glob.glob(path+'**/*.fits',recursive=True)
    files.sort()
    print(str(len(files))+" files.")
    for fi in files:
        file=fits.open(fi)
        cal_ver=file[0].header['CAL_VER']
        print(cal_ver, fi)

        
        
def obs_information_JWST(path):
    files = glob.glob(path+'**/*.fits',recursive=True)
    files.sort()
    print(str(len(files))+" files.")
    for fi in files:
        img = fits.open(fi)
        cal_ver = img[0].header['CAL_VER']
        name_fi = img[0].header['FILENAME']
        prog_id = img[0].header['PROGRAM']
        obs_nb  = img[0].header['OBSERVTN']
        obs_day = img[0].header['DATE-OBS']
        name_tar= img[0].header['TARGNAME']
        cat_tar = img[0].header['TARGCAT']
        filt    = img[0].header['FILTER']
        exp_ty  = img[0].header['EXP_TYPE']
        crds_ve = img[0].header['CRDS_CTX']
        exptime = img[0].header['DURATION']
        #print(name_fi, obs_day, obs_nb, cal_ver, crds_ve, exp_ty, filt, cat_tar)#, name_tar)
        #print(obs_nb, obs_day, cal_ver, crds_ve, name_fi, exp_ty, filt, cat_tar)#, name_tar)
        print(obs_nb, obs_day, cal_ver, crds_ve, exp_ty, filt, exptime, cat_tar, name_tar)
```

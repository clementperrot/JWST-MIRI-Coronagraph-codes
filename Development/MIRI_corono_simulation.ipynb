```python
from pandeia.engine.calc_utils import build_default_calc
from pandeia.engine.perform_calculation import perform_calculation
import matplotlib.pyplot as plt

```

```python


configuration = build_default_calc('jwst', 'miri', 'coronagraphy')
scene = {}
scene['position'] = {'x_offset': 0., 'y_offset': 0., 'orientation': 0., 'position_parameters': ['x_offset', 'y_offset', 'orientation']}
scene['shape'] = {'geometry': 'point'}
scene['spectrum'] = {'name': 'Phoenix Spectrum', 'spectrum_parameters': ['sed', 'normalization']}
scene['spectrum']['sed'] = {'sed_type': 'phoenix', 'key': 'g2v'}
scene['spectrum']['normalization'] = {'type': 'jwst', 'bandpass': 'miri,coronagraphy,f1140c', 'norm_flux': 2., 'norm_fluxunit': 'mjy'}
scene['spectrum']['lines'] = []
scene['spectrum']['extinction'] = {'bandpass': 'j', 'law': 'mw_rv_31', 'unit': 'mag', 'value': 0}
configuration['scene'][0] = scene

configuration['configuration']['scene_size'] = 10
configuration['configuration']['max_scene_size'] = 20
configuration['configuration']['dynamic_scene'] = True

configuration['configuration']['detector']['nexp'] = 1
configuration['configuration']['detector']['ngroup'] = 100
configuration['configuration']['detector']['nint'] = 1
configuration['configuration']['detector']['readout_pattern'] = 'fastr1'
configuration['configuration']['detector']['subarray'] = 'mask1140'

configuration['configuration']['instrument']['aperture'] = 'fqpm1140'
configuration['configuration']['instrument']['disperser'] = None
configuration['configuration']['instrument']['filter'] = 'f1140c'
configuration['configuration']['instrument']['instrument'] = 'miri'
configuration['configuration']['instrument']['mode'] = 'coronagraphy'

configuration['calculation']['noise']['cosmic_rays_on'] = True
configuration['calculation']['noise']['dark_current_on'] = True
configuration['calculation']['noise']['flat_field_noise_on'] = False
configuration['calculation']['noise']['readnoise_excess_additive_on'] = True
configuration['calculation']['noise']['readnoise_on'] = True
configuration['calculation']['noise']['scattering_on'] = False
configuration['calculation']['effects']['random_seed'] = -1
configuration['calculation']['effects']['saturation_on'] = True

configuration['scene'][0]['position']['x_offset'] = 0.
configuration['scene'][0]['position']['y_offset'] = 0.
configuration['scene'][0]['position']['orientation'] = 0.
configuration['scene'][0]['shape']['geometry'] = 'point'
configuration['scene'][0]['spectrum']['extinction']
configuration['scene'][0]['spectrum']['lines']
configuration['scene'][0]['spectrum']['name']
configuration['scene'][0]['spectrum']['normalization']
configuration['scene'][0]['spectrum']['sed']['sed_type'] = 'flat'
configuration['scene'][0]['spectrum']['sed']['unit'] = 'mjy'



report = perform_calculation(configuration)
print(report['scalar']['sn'])
```

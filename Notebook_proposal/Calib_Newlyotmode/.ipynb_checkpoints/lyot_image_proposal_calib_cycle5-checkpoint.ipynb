```python
import matplotlib, matplotlib.pyplot as plt
import stpsf
from astropy.io import fits
import numpy as np
import imodex_miri_postprocess_pipeline as imp #homemade pipeline
from matplotlib.colors import LogNorm
```

```python
lyot = stpsf.MIRI()
lyot.filter =  'F2300C'
lyot.pupil_mask = 'MASKLYOT' #'MASKLYOT'
lyot_psf = lyot.calc_psf(oversample=1)  
```

```python
stpsf.display_psf(lyot_psf,vmin=1e-8,vmax=1e-4,imagecrop=4) 
lyot_psf.writeto('/home/cperrot/Work/projet/JWST/Target/Calib/lyot_psf.fits', overwrite=True)
```

```python
fqpm = stpsf.MIRI()
fqpm.filter =  'F1140C'
fqpm.pupil_mask = 'MASKFQPM' #'MASKLYOT'
fqpm_psf = fqpm.calc_psf(oversample=1)  
```

```python
stpsf.display_psf(fqpm_psf,vmin=1e-8,vmax=1e-4,imagecrop=4) 
fqpm_psf.writeto('/home/cperrot/Work/projet/JWST/Target/Calib/fqpm_psf.fits', overwrite=True)
```

```python
miri = stpsf.MIRI()
miri.filter = 'F2300C'
miri.image_mask = 'LYOT2300' #'LYOT2300'
miri.pupil_mask = 'MASKLYOT' #'MASKLYOT'
coron_psf = miri.calc_psf()
```

```python
stpsf.display_psf(coron_psf,vmin=1e-8,vmax=1e-5) 
coron_psf.writeto('/home/cperrot/Work/projet/JWST/Target/Calib/stpsf.fits', overwrite=True)
```

```python
bkg_lyot = fits.open("/home/cperrot/Work/raw/JWST/Target/MAST/CAL/P01193/obs003/P01193_003_REF_19_PsA_background_dither1_cal.fits")
psf_lyot_im = fits.open("/home/cperrot/Work/projet/JWST/Target/Calib/lyot_psf.fits")
psf_fqpm_im = fits.open("/home/cperrot/Work/projet/JWST/Target/Calib/fqpm_psf.fits")
print(np.shape(bkg_lyot[1].data))
dim_bkg = np.shape(bkg_lyot[1].data)
dim_psf = np.shape(psf_lyot_im[0].data)
print(np.shape(psf_fqpm_im[0].data))
bkg_lyot[1].data = imp.clean_image_nan(bkg_lyot[1].data)
#psf_lyot_im[0].data = psf_lyot_im[0].data[432//2-304//2:432//2+304//2, 432//2-320//2:432//2+320//2]
#psf_fqpm_im[0].data = psf_fqpm_im[0].data[432//2-304//2:432//2+304//2, 432//2-320//2:432//2+320//2]
#print(np.shape(psf_lyot_im[0].data))
#print(np.shape(psf_fqpm_im[0].data))
img = np.copy(bkg_lyot[1].data)
img /= np.max(img)*0.5
lyot = np.copy(psf_lyot_im[0].data)
lyot /= np.max(lyot)
px=15
lyot = lyot[dim_psf[0]//2-px:dim_psf[0]//2+px, dim_psf[1]//2-px:dim_psf[1]//2+px]
img[167-px:167+px, 144-px:144+px] = np.log10(lyot)+1.5

fqpm = np.copy(psf_fqpm_im[0].data)
fqpm /= np.max(fqpm)
px=10
fqpm = fqpm[dim_psf[0]//2-px:dim_psf[0]//2+px, dim_psf[1]//2-px:dim_psf[1]//2+px]

img[167-px-round(7.5/0.11):167+px-round(7.5/0.11), 144-px-round(0.63/0.11):144+px-round(0.63/0.11)] = np.log10(fqpm)+1.5

pixscale = 0.11
scale=[(144)*pixscale,-1*(dim_bkg[1]-144)*pixscale, -1*(167)*pixscale,(dim_bkg[0]-167)*pixscale]

fig=plt.figure(figsize=(10,8))
plt.imshow(img, origin='lower',interpolation='none',extent=scale, vmin=0,vmax=1.5)#,norm=LogNorm(vmin=1e-2,vmax=2e0))
plt.xlabel("Arcsec")
plt.ylabel("Arcsec")
plt.show()
fig.savefig("/home/cperrot/Work/projet/JWST/Target/Calib/new_lyot_mode_illustration.png",format="png",bbox_inches='tight')
plt.close(fig)
```
